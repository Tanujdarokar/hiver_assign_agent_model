# Phase 1 & 3 Dataset Exploration & Intent Discovery
This notebook explores the Twitter Customer Support dataset for AppleSupport and visualizes intent distributions.

```python
import pandas as pd
import json
from pathlib import Path

ROOT = Path('..')
proc_df = pd.read_csv(ROOT / 'data' / 'processed' / 'apple_support_processed.csv')
print(f"Loaded {len(proc_df):,} AppleSupport conversation pairs.")
proc_df.head(5)
```

```python
with open(ROOT / 'data' / 'processed' / 'intents.json') as f:
    intents = json.load(f)
print(f"Loaded {len(intents)} intent definitions.")
for name, info in intents.items():
    print(f"{name:30s}: {info['approximate_frequency_pct']:.1f}%")
```
